# Aula 05 - Analise exploratoria de fraudes

```python
import pandas as pd
from pathlib import Path
import os

local = Path(r'F:\Kubeflow\Datasets\CreditCardFraud\creditcard.csv')
cluster = Path('/data/creditcard.csv')
arquivo = cluster if cluster.exists() else local
df = pd.read_csv(arquivo)
print('Dataset:', arquivo)
df.head()
```

```python
print('Registros:', len(df))
print('Colunas:', len(df.columns))
print('Valores ausentes:', df.isna().sum().sum())
```

```python
print('Distribuicao das classes:')
print(df['Class'].value_counts())
print('Percentual de fraude:', round(df['Class'].mean() * 100, 4))
```
