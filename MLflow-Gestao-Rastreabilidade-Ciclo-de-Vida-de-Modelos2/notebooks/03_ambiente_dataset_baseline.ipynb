```python
from pathlib import Path
import sys
raiz = Path.cwd()
if raiz.name == 'notebooks':
    raiz = raiz.parent
sys.path.insert(0, str(raiz))
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from src.aula03_ambiente_dataset import (
    gerar_base_credito,
    preparar_divisao,
    construir_pipeline,
    TARGET,
    FEATURES,
    NUM_COLS,
    CAT_COLS,
    RANDOM_STATE,
)
np.random.seed(RANDOM_STATE)
```

## Distribuição da variável inadimplente

```python
display(df[TARGET].value_counts())
display(df[TARGET].value_counts(normalize=True))
```

## Exploração: score interno e renda

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, coluna in zip(axes, ['score_interno', 'renda_mensal']):
    df.boxplot(column=coluna, by=TARGET, ax=ax)
    ax.set_title(coluna)
    ax.set_xlabel('inadimplente')
plt.suptitle('')
plt.tight_layout()
plt.show()
```
