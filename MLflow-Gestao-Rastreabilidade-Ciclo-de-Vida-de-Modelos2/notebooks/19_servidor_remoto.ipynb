# Aula 19 - Tracking Server remoto
Laboratorio cliente-servidor via HTTP em localhost.

```python
import json
from pathlib import Path
raiz = Path.cwd()
if raiz.name == 'notebooks': raiz = raiz.parent
arquivo = raiz / 'resultados/aula19_servidor_remoto.json'
resultado = json.loads(arquivo.read_text(encoding='utf-8'))
resultado
```

```python
print('TRACKING:', resultado['tracking_uri'])
print('RUN:', resultado['run_id'])
print('METRICAS:', resultado['metrics'])
print('ARTEFATO:', resultado['artifact_downloaded'])
```

## Arquitetura
- Cliente: Python / MLflow API
- Tracking Server: HTTP em 127.0.0.1:5050
- Backend Store: SQLite isolado
