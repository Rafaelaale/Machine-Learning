## Separação cronológica
O treino recebe o passado e o teste recebe as últimas 60 observações.

```python
tamanho_teste = 60
treino, teste = separar_treino_teste(
    serie_modelagem,
    tamanho_teste=tamanho_teste,
)
print('Tamanho de treino:', len(treino))
print('Tamanho de teste:', len(teste))
```

```python
print('Período de treino:')
print(treino['data_hora'].min(), 'até', treino['data_hora'].max())
print('Período de teste:')
print(teste['data_hora'].min(), 'até', teste['data_hora'].max())
print(diagnosticar_calendario(serie_modelagem, treino, teste))
```
