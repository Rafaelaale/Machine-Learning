```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from statsmodels.tsa.statespace.sarimax import SARIMAX
from src.aula13_treinamento_sarima import (
    carregar_dados,
    verificar_calendario,
    ajustar_sarima,
    prever_sarima,
    calcular_resultados,
    grafico_previsoes,
    grafico_mape,
    grafico_residuos,
    executar_aula,
)
treino, teste = carregar_dados()
print('Treino:', len(treino))
print('Teste:', len(teste))
```
